```python
# 1. Import Libraries

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 2. Loading the Dataset

titanic = pd.read_csv("train.csv")
```

```python
# 3. Quick Look into the Dataset

titanic.info()
```

```python
titanic.head()
```

```python
titanic.tail()
```

```python
# 4. Summary Statistics

titanic.describe()
```

```python
titanic.describe().transpose()
```

Basic information
1. How many people survived and died?

```python
titanic['Survived'].value_counts()
```

```python
sns.countplot(data=titanic, x='Survived')

plt.title('Number of Passengers Survived and Died')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()
```

2. How many passenger classes are there?

```python
titanic['Pclass'].value_counts()
```

```python
sns.countplot(data=titanic, x='Pclass')

plt.title('Passenger Class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.show()
```

3. How many genders are there?

```python
titanic['Sex'].value_counts()
```

```python
sns.countplot(data=titanic, x='Sex')

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.show()
```

4. Show passengers younger than 20

```python
less_than_20 = titanic[titanic['Age'] < 20]

less_than_20.head()
```

```python
print("Number of passengers younger than 20:", len(less_than_20))
```

5. Categories of passengers

```python
titanic['Sex'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Passenger Categories by Gender')
plt.ylabel('')
plt.show()
```

6. Cities of embarkation

```python
titanic['Embarked'].nunique()
```

```python
titanic['Embarked'].value_counts()
```

```python
titanic['Embarked'].value_counts().plot(kind='bar')

plt.title('Passengers by Embarkation Port')
plt.xlabel('Embarked Port')
plt.ylabel('Number of Passengers')
plt.show()
```

6. Missing Data

```python
titanic.isnull().sum()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    titanic.isnull(),
    yticklabels=False,
    cbar=False
)

plt.title('Missing Values in Titanic Dataset')
plt.show()
```

7. More Analysis

```python
#Survival by Gender
sns.countplot(
    data=titanic,
    x='Survived',
    hue='Sex'
)

plt.title('Survival by Gender')
plt.show()
```

```python
#Survival by Passenger Class
sns.countplot(
    data=titanic,
    x='Survived',
    hue='Pclass'
)

plt.title('Survival by Passenger Class')
plt.show()
```

```python
#Age Distribution
titanic['Age'].plot(
    kind='hist',
    bins=30
)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
#Age vs Fare by Class
sns.scatterplot(
    data=titanic,
    x='Age',
    y='Fare',
    hue='Pclass'
)

plt.title('Age vs Fare by Passenger Class')
plt.show()
```

```python
#Age vs Fare by Gender
sns.scatterplot(
    data=titanic,
    x='Age',
    y='Fare',
    hue='Sex'
)

plt.title('Age vs Fare by Gender')
plt.show()
```

8. Correlation

```python
correlation = titanic.select_dtypes(
    include=np.number
).corr()

correlation
```

```python
correlation['Survived']
```

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='crest'
)

plt.title('Correlation Heatmap')
plt.show()
```
